## Boolean comparisons and masking

A comparison applied to a NumPy array produces a **Boolean mask**: one `True` or `False` per element. Combine conditions using `&` (and), `|` (or), and `~` (not). Put parentheses around each comparison.


```python
measurements = np.array([12, 18, 25, 31, 42, 15, 36])
upper = 30  # Values above this limit satisfy the condition.
mask = measurements > upper
print("Mask:", mask)
print("Selected values:", measurements[mask])
print("Count:", mask.sum())
```

```python
lower = 15
selected = (measurements < lower) | (measurements > upper)
print(measurements[selected])
print("Values inside range:", measurements[~selected])
```

## Pandas Series and DataFrames

A **Series** has values and an index. A **DataFrame** is a table with named columns. Boolean masks can filter rows; `.loc` selects by labels and `.iloc` by integer positions.


```python
import pandas as pd

df = pd.DataFrame({
    "temperature": [19.0, 21.5, 24.0, 18.5, 23.0],
    "flow": [5.2, 5.8, 6.1, 5.4, 6.0]
}, index=pd.date_range("2026-01-01", periods=5, freq="h"))
display(df)
print(df["temperature"])  # Series
print(df.iloc[0, 0])      # First row, first column
```

```python
warm = df["temperature"] > 21
print("Boolean mask:")
print(warm)
print("Filtered rows:")
display(df.loc[warm])
display(df.loc[(df["temperature"] > 20) & (df["flow"] < 6)])
```

## Missing values and basic data preparation

`NaN` represents missing numerical data. Use `.isna()` to identify missing entries, `.dropna()` to remove them, and `.fillna()` to replace them. Missing-value treatment depends on the purpose of the analysis.


```python
sample = pd.Series([2.0, np.nan, 4.0, 5.0], name="sample")
print(sample.isna())
print("Missing values:", sample.isna().sum())
print("Without missing values:", sample.dropna().to_list())
print("Filled with the mean:", sample.fillna(sample.mean()).to_list())
```

## Plot selected observations

The same Boolean mask used for filtering can select points to highlight on a plot.


```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 3.5))
ax.plot(df.index, df["temperature"], label="Temperature", color="#2C5282")
selected = df["temperature"] > 21
ax.scatter(df.index[selected], df.loc[selected, "temperature"],
           marker="o", color="#D62728", label="Selected observations")
ax.set_ylabel("Temperature")
ax.legend()
fig.autofmt_xdate()
plt.show()
```

### Plotly

Plotly is an interactive visualization library. Unlike Matplotlib, Plotly figures allow users to zoom, pan, and inspect individual data points by hovering over them. The underlying data selection remains exactly the same; only the plotting library changes.

```python
import plotly.graph_objects as go

fig = go.Figure()

# Plot all observations as a line.
fig.add_trace(go.Scatter(
    x=df.index,
    y=df["temperature"],
    mode="lines",
    name="Temperature",
    line=dict(color="#2C5282")
))

# Highlight selected observations using the same Boolean mask.
selected = df["temperature"] > 21

fig.add_trace(go.Scatter(
    x=df.index[selected],
    y=df.loc[selected, "temperature"],
    mode="markers",
    name="Selected observations",
    marker=dict(color="#D62728", size=9)
))

# Configure the figure layout.
fig.update_layout(
    title="Temperature measurements",
    xaxis_title="Time",
    yaxis_title="Temperature",
    template="plotly_white",
    width=850,
    height=400,
    legend=dict(x=1.02, y=1, xanchor="left", yanchor="top"),
    margin=dict(r=170)
)

fig.show()
```
